```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
import os
import time

# ==========================================
# 1. Configuration
# ==========================================
# Target data file check
results_path = ".."
target_csv = os.path.join(results_path,'resultados', 'consumo_previsto_todos_cenarios_2026_2035.csv')

# List of notebooks to be executed in sequence after simulation execution in Gama
scripts = [ "Script_soma_consumo120m_por_mes.ipynb","Script_comparacao_mensal_consumo_cenarios_lstm_sarima.ipynb",
    "Script_perfil_consumo_por_setor.ipynb", "Script_conexoes_por_bairro.ipynb",
    "Script_comparacao_mensal_consumo_cenarios_lstm_sarima_labels.ipynb","Script_projecao_comparativo_diferencas_cenarios_sarima.ipynb",
    "Script_analise_distribuicao.ipynb", "Script_analise_parametrica_cenarios.ipynb", 
    "Script_projecao_comparativo_barras_diferencas_cenarios_sarima.ipynb",
    "Script_boxsplot_comparacao_cenarios.ipynb", "Script_mapa_calor_impacto_cenarios_horizontal.ipynb",
    "Script_projecao_setores_mais_afetados_pos_simulacao.ipynb", 
     "Script_analise_mudanca_perfil.ipynb", "Script_setores_mais_afetados_pos_simulacao_detalhes.ipynb"
]

# ==========================================
# 2. Notebook Execution Pipeline
# ==========================================
print("="*60)
print("Processing Notebooks Initial Pipeline")
print("="*60)

if not os.path.exists(target_csv):
    print(f"Error: Required data file {os.path.basename(target_csv)} not found.")
    print("Please ensure the simulation results are available before running.")
else:
    print(f"Data detected. Processing {len(scripts)} notebooks...\n")
    
    # Configure the processor 
    processor = ExecutePreprocessor(timeout=1500, kernel_name='python3')

    for script in scripts:
        if os.path.exists(script):
            start_clock = time.time()
            print(f"Executing: {script}...", end=" ", flush=True)
            try:
                # Load notebook content
                with open(script, 'r', encoding='utf-8') as f:
                    nb_content = nbformat.read(f, as_version=4)
                
                # Execute notebook and generate outputs
                processor.preprocess(nb_content, {'metadata': {'path': os.getcwd()}})
                
                # Save the notebook with updated results/cells
                with open(script, 'w', encoding='utf-8') as f:
                    nbformat.write(nb_content, f)
                
                end_clock = time.time() - start_clock
                print(f"Done ({end_clock:.1f}s)")
                
            except Exception as error:
                print(f"Failed: {str(error)}")
        else:
            print(f"Warning: File {script} not found in the directory.")

print("\n" + "="*60)
print("Pipeline execution completed.")
print("="*60)
```
