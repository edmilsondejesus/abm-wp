```python
import pandas as pd
import numpy as np
import os
from statsmodels.tsa.statespace.sarimax import SARIMAXResults
from datetime import datetime

def main():
    print("--- Starting Monthly Consumption Projection (2016-2025) ---")

    MODEL_LOAD_PATH = 'best_sarima_model.pkl'
    OUTPUT_CSV = 'previsao_consumo_sarima_2016_2025.csv'

    if not os.path.exists(MODEL_LOAD_PATH):
        print(f"Error: Trained model not found at {MODEL_LOAD_PATH}")
        return

    # ==========================================
    # 1. Loading the Best Model
    # ==========================================
    model_results = SARIMAXResults.load(MODEL_LOAD_PATH)
    print("Model loaded successfully.")

    # ==========================================
    # 2. Generating Projection Range
    # ==========================================
    # Target range: Jan/2016 to Dec/2025
    start_date = "2016-01-01"
    end_date = "2025-12-01"
    
    # We use the model to predict for the specific range
    # Note: SARIMA can predict historical values (fit) or future values (forecast)
    projection = model_results.get_prediction(start=start_date, end=end_date)
    mean_projection = projection.predicted_mean

    # ==========================================
    # 3. Structuring and Saving Results
    # ==========================================
    results_df = pd.DataFrame({
        'Date': mean_projection.index,
        'Projected_Consumption': mean_projection.values
    })

    # Ensure format consistency
    results_df['Date'] = results_df['Date'].dt.strftime('%Y-%m')
    
    results_df.to_csv(OUTPUT_CSV, index=False)
    
    print(f"\nProjection completed!")
    print(f"Results saved to: {OUTPUT_CSV}")
    print("\nSample of results:")
    print(results_df.head())
    print("...")
    print(results_df.tail())

if __name__ == "__main__":
    main()
```
