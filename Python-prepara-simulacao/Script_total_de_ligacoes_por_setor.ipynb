```python
import pandas as pd
import os

# Define paths
path = '..' # Last one active

# 1. Load the DE-PARA mapping file
mapping_df = pd.read_csv(os.path.join(path, 'resultados', 'de_para_setores_consumidores.csv'), sep=';', encoding='utf-8-sig')
sector_mapping = dict(zip(mapping_df['SETOR_ORIGINAL'].astype(str), mapping_df['SETOR_SIMPLIFICADO']))

# 2. Load the .csv file with CD_SETOR and SK_MATRICULA columns
df = pd.read_csv(os.path.join(path, 'includes', 'Tabela_consumidores_Itapua_com_setor.csv'), sep=',')

# 3. Convert CD_SETOR using the mapping file
df['CD_SETOR'] = df['CD_SETOR'].astype(str).map(sector_mapping)

# 4. Quantify the number of connections (SK_MATRICULA) per sector (CD_SETOR)
connections_by_sector = df.groupby('CD_SETOR')['SK_MATRICULA'].count().reset_index()
connections_by_sector.rename(columns={'SK_MATRICULA': 'NUM_CONNECTIONS'}, inplace=True)

# 5. Sort sectors by number of connections (from highest to lowest)
connections_by_sector = connections_by_sector.sort_values(by='NUM_CONNECTIONS', ascending=False)

# 6. Select the top 10 sectors
top_10_sectors = connections_by_sector.head(10)

# 7. Save the result to a new CSV file (optional)
top_10_sectors.to_csv(os.path.join(path, 'resultados', 'Tabela_top_10_setores.csv'), index=False)

# 8. View the result
print(top_10_sectors)
```
